## 3. Success metric

*One metric you can defend. What number means 'good'?*

The main success metric would be **Precision@K**, where K represents the number of pages the team can realistically review in a given period.

For example, Precision@50 would measure how many of the top 50 prioritized pages turn out to be useful refresh candidates.

This metric fits the business action because the team is likely to work on a limited number of pages first. A high Precision@K means the system is putting more useful opportunities near the top of the list, rather than wasting the team's review time on low-value pages.

I would also consider recall and ranking quality during evaluation, but Precision@K is the most directly connected to the prioritization decision.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could prioritize pages using a simple threshold, such as selecting every page with search volume above a certain value. However, a single threshold may miss useful opportunities because page performance depends on multiple signals.

An ML-based approach can learn patterns across several features at the same time, such as search volume, impressions, clicks, content age, average position, and engagement. This can produce a more flexible ranking of pages than one manually chosen rule.

ML is useful here only if it improves the prioritization decision on unseen data. If a simple fixed rule performs just as well, then the rule may be preferable because it is easier to understand and maintain.
